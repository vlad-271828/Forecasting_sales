# Соединяем станции и продажи
В записях подготовленной нами таблицы про продажи есть информация о том, какая станция делала продажи, а также у нас есть таблица со станциями и информацией о них. Первую таблицу нужно дополнить второй (именно с этой таблицей мы в итоге и будем работать).

```python
import pandas as pd


stations = pd.read_csv('../../data/Cleared_data/stations.csv')
sales = pd.read_csv('../../data/Cleared_data/sales.csv')
```

```python
sales
```

```python
stations
```

```python
sales_stations = sales.merge(stations, left_on='station_id', right_on='id', how='left')
sales_stations = sales_stations.drop(['id'], axis=1)
```

```python
sales_stations.to_csv('../../data/Cleared_data/sales_stations.csv', index=False)
```
